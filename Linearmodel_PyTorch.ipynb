```python
import pandas as ps
import numpy as np
import torch
import torch as nn
import matplotlib.pyplot as plt
```

```python

weight,bias = 0.7,0.3
X = torch.arange(0,100,0.5).unsqueeze(dim=1)
X[:5],len(X)
```

```python
y = weight * X + bias
y[:5].squeeze()
```

```python
X.shape,y.shape
```

```python
# train test split
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2, random_state=2)
```

```python
len(X_train),len(y_test)
```

```python
def plot_train_test(train_data= X_train,
                    train_labels =y_train,
                    test_data =X_test,
                    test_labels= y_test,
                    predictions = None):
    plt.scatter(train_data,train_labels,s=10, c='red',label='Training')
    plt.scatter(test_data,test_labels,s=10, c='blue', label='testing')


    #if plot_train_test is not None:
        #  plt.scatter(X_test, predictions)
plot_train_test();
```

```python
import torch.nn as nn
X_train.shape
```

```python
class Linearmodel(nn.Module):
      def __init__(self):
          super().__init__()
          self.layer = nn.Linear(in_features=1 , out_features=1)
          #self.layer2 = nn.Linear(in_features=5, out_features=1)

      def forward(self,x):
          # return self.layer2(self.layer(x))
            return self.layer(x)

model1 = Linearmodel()
model1
```

```python

#torch.manual_seed(4)
model1.state_dict()
```

## loss function

```python
loss_fn = nn.L1Loss()
optimizer = torch.optim.SGD(params= model1.parameters(), lr=0.01)
```

## training Loop

```python
torch.manual_seed(42)

epochs= 200

for epoch in range(epochs):
    ## Training
    model1.train()
    #1. Forward pass
    y_pred = model1(X_train)
    # calculate the loss
    loss = loss_fn( y_pred, y_train)
    # zero gradient
    optimizer.zero_grad()
    # back propagation
    loss.backward()
    # step
    optimizer.step()

    if epoch % 20 == 0:
       model1.eval()
       with torch.inference_mode():
            test_pred = model1(X_test)
            test_loss = loss_fn(test_pred, y_test)
       print(f'Epoch: {epoch}| Train loss: {loss} |Test loss : {test_loss}')
```
